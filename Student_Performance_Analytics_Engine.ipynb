```python
ort numpy as np
import pandas as pd
from google.colab import files

# Set random seed for reproducibility
np.random.seed(42)

# Generate a 500x6 marks matrix (500 students, 6 subjects) with scores ranging between 35 and 100
marks_matrix = np.random.randint(35, 101, size=(500, 6))

print("Marks Matrix Shape:", marks_matrix.shape)
print("\nFirst 5 students' marks preview:\n", marks_matrix[:5])

# --- Axis 0: Subject-wise Statistics (Calculated down the columns) ---
subject_sums = np.sum(marks_matrix, axis=0)
subject_means = np.mean(marks_matrix, axis=0)
subject_medians = np.median(marks_matrix, axis=0)
subject_stds = np.std(marks_matrix, axis=0)
subject_vars = np.var(marks_matrix, axis=0)

print("--- Subject-Wise Statistics (Axis = 0) ---")
for i in range(6):
    print(f"Subject {i+1} -> Mean: {subject_means[i]:.2f} | Median: {subject_medians[i]:.1f} | Std: {subject_stds[i]:.2f} | Var: {subject_vars[i]:.2f}")

# --- Axis 1: Student-wise Statistics (Calculated across the rows) ---
student_sums = np.sum(marks_matrix, axis=1)
student_means = np.mean(marks_matrix, axis=1)
student_medians = np.median(marks_matrix, axis=1)
student_stds = np.std(marks_matrix, axis=1)
student_vars = np.var(marks_matrix, axis=1)

print("\n--- First 5 Students' Summary (Axis = 1) ---")
for s in range(5):
    print(f"Student {s+1} -> Total: {student_sums[s]} | Mean: {student_means[s]:.2f} | Std: {student_stds[s]:.2f}")

    # --- Z-Scores via Broadcasting ---
# Formula: Z = (X - mean) / std for each subject (axis=0)
# NumPy automatically broadcasts the 1D mean and std arrays across the 500x6 matrix rows
mean_broadcast = np.mean(marks_matrix, axis=0)
std_broadcast = np.std(marks_matrix, axis=0)
z_scores = (marks_matrix - mean_broadcast) / std_broadcast

# --- Outlier Detection ---
# Flag any student who has an absolute Z-score greater than 3 in any subject
outlier_mask = np.abs(z_scores) > 3.0
students_with_outliers = np.any(outlier_mask, axis=1)
num_outliers = np.sum(students_with_outliers)

print(f"Number of students flagged as outliers (|Z| > 3 in any subject): {num_outliers}")

# --- Student Rankings ---
# Rank students based on total marks in descending order (Rank 1 = highest total)
sorted_indices = np.argsort(-student_sums)
ranks = np.empty_like(sorted_indices)
ranks[sorted_indices] = np.arange(1, 501)

print("\n--- Top 5 Students by Rank ---")
for r in range(1, 6):
    student_idx = sorted_indices[r-1]
    print(f"Rank {r}: Student Index {student_idx} (Total Marks: {student_sums[student_idx]})")

    # --- Combine Original Matrix with Metrics using hstack ---
# Reshape 1D arrays into column vectors (shape: 500x1)
col_totals = student_sums.reshape(-1, 1)
col_means = student_means.reshape(-1, 1)
col_ranks = ranks.reshape(-1, 1)
col_outliers = students_with_outliers.astype(int).reshape(-1, 1)

# Horizontally stack marks and analytics columns together
enhanced_matrix = np.hstack((marks_matrix, col_totals, col_means, col_ranks, col_outliers))

# --- Export to CSV ---
columns = [
    'Subject_1', 'Subject_2', 'Subject_3', 'Subject_4', 'Subject_5', 'Subject_6',
    'Total_Marks', 'Average_Marks', 'Rank', 'Is_Outlier'
]

df_results = pd.DataFrame(enhanced_matrix, columns=columns)

csv_filename = 'student_performance_analytics.csv'
df_results.to_csv(csv_filename, index=False)
print(f"Successfully created and saved DataFrame to {csv_filename}")



```
